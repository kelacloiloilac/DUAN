# 01 - Chuẩn bị dữ liệu

Tạo split train/validation/test theo stratification. Test được khóa lại cho notebook đánh giá cuối.

```python
from pathlib import Path
import sys
PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(PROJECT_ROOT))
from src.data import discover_images, seed_everything, stratified_split
DATA_ROOT = PROJECT_ROOT.parent / 'PlantVillage-Dataset-master' / 'PlantVillage-Dataset-master' / 'raw' / 'color'
OUTPUT = PROJECT_ROOT / 'data' / 'metadata.csv'
SEED = 42
seed_everything(SEED)
metadata = stratified_split(discover_images(DATA_ROOT), seed=SEED, val_size=0.15, test_size=0.15)
OUTPUT.parent.mkdir(exist_ok=True)
metadata.to_csv(OUTPUT, index=False)
metadata.groupby(['split', 'class_name']).size().unstack(fill_value=0)
```

```python
assert not metadata['path'].duplicated().any()
assert set(metadata['split']) == {'train', 'val', 'test'}
print(metadata.groupby('split').size())
print(f'Saved to {OUTPUT}')
```
